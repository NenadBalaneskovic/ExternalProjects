# 0. Project Ideas

I would like to design a PyQt 5 GUI which lets the user choose 1) an n-gonal sensor housing structure (for n in a slider-range from 3 to 21), 2) stress (force) applied to each of the n corners in a slider-range from 1 N to 100 N, and 3) heat intensity in a slider range from 1°C to 100 °C. Then the gui should visualize the n-gonal sensor housing shape as a rod-structure including the selected n force vectors and n heat-sources in a graphics window. After pressing the "Robustness analysis" button the gui should perform 1) a static rod-work deformation analysis and identify vulnerable housing parts, 2) an FEM analysis of stress and heat propagation through the n-gonal sensor housing and visualize the results in a second graphics window. In addition, the gui should store all graphical results from both graphics windows as well as all plots and quantitative results in an XAI-driven well-explained pdf report. The aim would converge towards showcasing a well-conceptualized user-interface for simulation of sensor-housing stress test scenarios. How could we brainstorm this project?

This project combines **PyQt5 GUI design**, **simulation visualization**, **FEM analysis**, and **XAI-driven reporting** into one narratable, consulting-ready asset.

### `gui/event_handlers.py`
```python
class EventHandlers:
    def __init__(self, sidebar, visualization):
        self.sidebar = sidebar
        self.visualization = visualization
        self.connect_events()

    def connect_events(self):
        self.sidebar.run_button.clicked.connect(self.run_analysis)

    def run_analysis(self):
        n = self.sidebar.n_slider.value()
        force = self.sidebar.force_slider.value()
        heat = self.sidebar.heat_slider.value()
        fem = self.sidebar.fem_checkbox.isChecked()

        # Placeholder: connect to simulation
        self.visualization.rod_label.setText(f"Rod structure for n={n}, force={force}N, heat={heat}°C")
        if fem:
            self.visualization.fem_label.setText("Running FEM analysis...")
        else:
            self.visualization.fem_label.setText("Running analytical rod analysis...")
```

```python
        # Add widgets to layout
        layout.addWidget(self.sidebar, stretch=1)
```

```python
import math


def run_rod_analysis(n, force, heat):
    """
    Perform a simplified analytical rod deformation analysis for an n-gonal sensor housing.

    Args:
        n (int): Number of polygon corners (3–21).
        force (int): Force applied at each corner [N].
        heat (int): Heat intensity applied at each corner [°C].

    Returns:
        dict: Results including deflections, max deflection, and critical corner index.
    """

    # Store deflections for each corner
    deflections = []

    for i in range(n):
        # Angle of the corner in radians
        angle = (2 * math.pi * i) / n

        # Simplified deformation model:
        # Deflection = force * sin(angle) + heat contribution
        deflection = (force * math.sin(angle)) + (heat * 0.01)
        deflections.append(deflection)

    # Identify maximum deflection and critical corner
    max_deflection = max(deflections, key=abs)
    critical_corner = deflections.index(max_deflection)

    # Package results
    results = {
        "mode": "analytical",
        "n": n,
        "force": force,
        "heat": heat,
        "deflections": deflections,
        "max_deflection": max_deflection,
        "critical_corner": critical_corner,
    }

    return results
```

```python
import math


def generate_polygon_vertices(n, radius=1.0):
    """
    Generate vertices of a regular n-gon polygon.
    Args:
        n (int): Number of polygon corners (3–21).
        radius (float): Radius of circumscribed circle.
    Returns:
        list of tuples: (x, y) coordinates of vertices.
    """
    vertices = []
    for i in range(n):
        angle = (2 * math.pi * i) / n
        x = radius * math.cos(angle)
        y = radius * math.sin(angle)
        vertices.append((x, y))
    return vertices


def generate_polygon_edges(vertices):
    """
    Generate edges connecting consecutive vertices of the polygon.
    Args:
        vertices (list): List of (x, y) coordinates.
    Returns:
        list of tuples: (vertex_index_start, vertex_index_end).
    """
    edges = []
    n = len(vertices)
    for i in range(n):
        edges.append((i, (i + 1) % n))  # wrap around to close polygon
    return edges


def generate_polygon_mesh(n, radius=1.0):
    """
    Generate a simple polygon mesh structure for FEM analysis.
    Args:
        n (int): Number of polygon corners.
        radius (float): Radius of circumscribed circle.
    Returns:
        dict: Mesh data including vertices and edges.
    """
    vertices = generate_polygon_vertices(n, radius)
    edges = generate_polygon_edges(vertices)

    mesh = {
        "n": n,
        "vertices": vertices,
        "edges": edges,
    }
    return mesh
```

```python
def plot_rod_structure(n, force, heat):

    Args:
        n (int): Number of polygon corners (3–21).
        force (int): Force applied at each corner [N].
        heat (int): Heat intensity applied at each corner [°C].

    Returns:
        dict: Visualization data including vertices, forces, and heat markers.
    """

    vertices = []
    for i in range(n):
        angle = (2 * math.pi * i) / n
        x = math.cos(angle)
        y = math.sin(angle)
        vertices.append((x, y))

    # Force vectors: outward arrows from each vertex
    forces = []
    for (x, y) in vertices:
        # Normalize vector and scale by force
        length = math.sqrt(x**2 + y**2)

        forces.append((fx, fy))

    heat_sources = [heat for _ in range(n)]

    visualization_data = {
        "vertices": vertices,
        "forces": forces,
        "heat_sources": heat_sources,
        "summary": f"Rod structure with {n} corners, force={force}N, heat={heat}°C"
    }

    return visualization_data
```

```python
# main.py

import sys
from PyQt5.QtWidgets import QApplication
from gui.main_window import MainWindow
import config


def main():
    """
    Entry point for the Sensor Stress Analyzer application.
    Initializes the PyQt5 application and launches the main window.
    """

    app = QApplication(sys.argv)
    app.setApplicationName(config.APP_TITLE)

    # Create and show main window
    window = MainWindow()

    main()
```
