```python
from fpdf import FPDF
import os

def create_test_documents(output_dir="test_docs"):
    os.makedirs(output_dir, exist_ok=True)

    def make_pdf(filename, lines):
        pdf = FPDF()
        pdf.add_page()
        pdf.set_font("Arial", size=12)
        for line in lines:
            # Replace unsupported characters
            safe_line = line.replace("€", "EUR").replace("–", "-")
            pdf.multi_cell(0, 10, safe_line)
        pdf.output(os.path.join(output_dir, filename))


    # 📄 Invoice (Clean)
    make_pdf("invoice_clean.pdf", [
        "Rechnungsnummer: INV-DE-2025-0001",
        "Rechnungsdatum: 01 Oktober 2025",
        "Verkauft von: Heinz Guenter Walsdorf",
        "Zahlbetrag: 9,82 EUR",
        "Beschreibung: 5 Druckerpatronen kompatibel zu CLI-526GY",
        "USt-IDNr. DE244348108"
    ])

    # ⚠️ Invoice (Tampered)
    make_pdf("invoice_tampered.pdf", [
        "Rechnungsnummer: INV-DE-2025-0001",
        "Rechnungsdatum: 01 Oktober 2025",
        "Verkauft von: Heinz Günter W@lsdorf",
        "Zahlbetrag: 9,82 EUR",
        "Beschreibung: 5 Druckerpatronen kompatibel zu CLI-526GY",
        "Hinweis: Dieses Dokument wurde manipuliert.",
        "Font mismatch detected."
    ])

    # 📄 Receipt (Clean)
    make_pdf("receipt_clean.pdf", [
        "Quittung für Zahlung",
        "Betrag: 9,82 EUR",
        "Zahlungsdatum: 02 Oktober 2025",
        "Transaktions-ID: TXN-302-2025-0001",
        "Zahlungsmethode: Kreditkarte"
    ])

    # ⚠️ Receipt (Tampered)
    make_pdf("receipt_tampered.pdf", [
        "Quittung für Zahlung",
        "Betrag: 12,00 EUR",
        "Zahlungsdatum: 02 Oktober 2025",
        "Transaktions-ID: TXN-302-2025-0001",
        "Zahlungsmethode: Unbekannt",
        "Hinweis: Korrigierte Quittung – ursprünglicher Betrag abweichend."
    ])

    # 📄 Other Document (Clean)
    make_pdf("delivery_note_clean.pdf", [
        "Lieferschein",
        "Lieferadresse: Zorica Henne, Seehofstraße 12a, Frankfurt",
        "Lieferdatum: 03 Oktober 2025",
        "Artikel: Druckerpatronen CLI-526GY",
        "Unterschrift: Empfänger bestätigt Erhalt"
    ])

    # ⚠️ Other Document (Tampered)
    make_pdf("delivery_note_tampered.pdf", [
        "Lieferschein",
        "Lieferadresse: Zorica Henne, Seehofstraße 99X, Frankfurt",
        "Lieferdatum: 03 Oktober 2025",
        "Artikel: Druckerpatronen CLI-526GY",
        "Hinweis: Unzustellbar – Adresse geändert.",
        "Unterschrift fehlt"
    ])

    print(f"Test documents created in: {output_dir}")
```
