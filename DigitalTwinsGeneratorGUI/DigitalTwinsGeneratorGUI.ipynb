# 🔄 **Interaction Between Generator and Analyzer GUIs**

## **Step-by-step Workflow**

### **1. Start the Generator GUI**
- We configure:
  - Number and type of columns (checkboxes)
  - Target file size or row count
  - Output format (CSV or Parquet)
  - Sampling frequency and duration
- We click **“Generate”**
- The generator begins writing telemetry data to a file:
  - Either as a growing CSV
  - Or as chunked Parquet blocks
- The file is continuously updated on disk (e.g. every few seconds or MB)

---

### **2. Start the Analyzer GUI**
- We select:
  - The same file path (CSV or Parquet)
  - Refresh interval (e.g. every 5 seconds)
  - Columns to monitor
  - Analysis modules to activate (clustering, forecasting, NLP, etc.)
- We click **“Start”**
- The analyzer enters a loop:
  - Every N seconds, it checks the file for new data
  - It reads only the new rows (tail mode) or re‑scans (lazy mode)
  - It updates plots, metrics, alerts, and logs

---

## **How They Stay Decoupled Yet Synchronized**

| Aspect | Generator GUI | Analyzer GUI |
|--------|----------------|----------------|
| File creation | Writes to disk | Reads from disk |
| Timing | Simulates real-time | Polls at intervals |
| Format | CSV or Parquet | Auto-detects format |
| Schema | User-defined | Dynamically adapts |
| Communication | Indirect (via file) | Indirect (via file) |

There’s **no direct socket, pipe, or messaging** between them — the file is the shared medium. This keeps both GUIs simple, testable, and modular.

---

## **Best Practices for Smooth Interaction**

- **File locking:** Generator should flush and close each write chunk cleanly so the analyzer can read without conflict.
- **Timestamp column:** Include it so the analyzer can filter by time.
- **Cycle counter or row ID:** Helps the analyzer detect new rows.
- **Chunk size:** Generator should write in chunks (e.g. every 10,000 rows) to avoid RAM overload and allow analyzer to catch up.
- **Analyzer tail logic:** Use Polars or Pandas to read only the last N rows or rows since last timestamp.

---

## **Optional Enhancements Later**

- **Live preview in generator:** So you can see the stream as it’s written.
- **Progress bar in analyzer:** Shows how much of the file has been processed.
- **Shared config file:** So both GUIs know the schema and sampling rate.
- **Alert file or socket (optional):** If you ever want real-time alerts from analyzer to generator (e.g. “stop simulation, anomaly detected”).

# 🛠️ **How the Generator Writes `config.json`**

Inside `core/config_writer.py`:

```python
import json
from datetime import datetime

def write_config(config, schema, output_path="config.json"):
    data = {
        "generator": {
            "version": "1.0",
            "timestamp": datetime.utcnow().isoformat()
        },
        "output": {
            "file_path": config.file_path,
            "file_format": config.file_format,
            "estimated_size_gb": config.target_size_gb,
            "chunk_size_rows": config.chunk_size_rows
        },
        "schema": {
            "columns": schema
        },
        "sampling": {
            "frequency_hz": config.frequency_hz,
            "interval_ms": int(1000 / config.frequency_hz)
        },
        "timestamp": {
            "enabled": config.timestamp_enabled,
            "format": config.timestamp_format
        },
        "preview": {
            "enabled": config.preview_enabled,
            "buffer_seconds": config.preview_buffer_seconds
        },
        "alerts": {
            "socket_enabled": config.socket_enabled,
            "socket_host": config.socket_host,
            "socket_port": config.socket_port
        }
    }

    with open(output_path, "w") as f:
        json.dump(data, f, indent=4)
```
