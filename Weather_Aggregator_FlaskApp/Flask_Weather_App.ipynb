## 🔌 Cost-Free Hugging Face Integration
- **Serverless Inference API**: Hugging Face offers free-tier access to thousands of models via hosted endpoints — no local downloads required.
- **How it works**:
  - You select models from the Hugging Face Hub that support inference.
  - Authenticate using an API token.
  - Send requests via HTTP POST with input payloads.
- **Best practices**:
  - Use lightweight models for faster response.
  - Batch or parallelize requests to reduce latency.
  - Monitor usage to stay within free-tier limits.

## 🧠 Ensemble Size & Meta-Learner Strategy

### ✅ Ensemble Size
- **30 models** is a solid minimum for robust meta-learning.
- Consider diversity: mix autoregressive LLMs, transformers, time-series regressors, and physics-inspired models.
- Use model confidence bands as features for the meta learner.

### 🔁 Meta-Learner Options
| Meta Learner        | Pros                                  | Cons                                  | Use Case                          |
|---------------------|----------------------------------------|---------------------------------------|-----------------------------------|
| **Logistic Regression** | Interpretable, fast                   | Limited expressiveness                | Binary classification (e.g., rain/no rain) |
| **Linear Regression**   | Simple, interpretable                | Poor with nonlinearities              | Aggregating temperature forecasts |
| **Ridge/Lasso**         | Regularized, avoids overfitting      | Still linear                          | High-dimensional ensemble inputs  |
| **Gradient Boosting**   | High accuracy, handles interactions  | Less interpretable                    | Complex ensemble fusion           |
| **Stacked Neural Net**  | Flexible, nonlinear                  | Requires tuning, slower               | Deep ensemble fusion              |

**Recommendation**: Start with **ridge regression** for interpretability and robustness, then experiment with **gradient boosting** or **shallow neural nets** for performance.

## ⏱️ Calculation Time Estimates
- **Local Flask orchestration**: negligible overhead.
- **Weather API call**: ~0.5–1.5 seconds.
- **Hosted model inference**:
  - Lightweight models: ~0.5–2 seconds per call.
  - Heavy models (e.g., LLaMA, GPT-style): ~3–10 seconds.
- **Meta learner aggregation**: ~0.5 seconds locally.
- **Total latency** (parallelized): **~3–6 seconds** for full ensemble + forecast.

## 🧩 Optimization Tips
- Use **async Flask routes** to parallelize model calls.
- Cache frequent locations and model outputs.
- Precompute ensemble weights for common cities.
- Use **confidence-weighted voting** or **Bayesian averaging** to reduce runtime.

### 2. **Helper Function**

```python
# Cell 2: Hugging Face inference helper
def query_hf_model(model_name, payload):
    response = requests.post(
        HF_API_URL + model_name,
        headers=headers,
        json=payload
    )
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error {response.status_code}: {response.text}")
        return None
```

```python
def ensemble_forecast(input_data, models=MODEL_LIST):
    results = {}
    for model in models:
        output = query_hf_model(model, {"inputs": input_data})
        results[model] = output
    return results
```

```python
# Cell 5: Meta learner aggregation
from sklearn.linear_model import Ridge

def meta_aggregate(local_forecasts, hf_results):
    # Convert Hugging Face outputs into numeric features (stubbed here)
    hf_features = [len(str(v)) if v else 0 for v in hf_results.values()]
    
    # Combine local + HF features
    X = np.array([local_forecasts + hf_features])
    y = np.mean(local_forecasts)  # target proxy
    
    ridge = Ridge(alpha=1.0)
    ridge.fit(X, [y])
    return ridge.predict(X)
```

```python
Handles inference requests, error management, and ensemble-ready outputs.
"""

import requests
import logging

# Configure logging for governance
logger = logging.getLogger(__name__)
logger.setLevel(logging.INFO)

class HuggingFaceModelClient:
    def __init__(self, api_token: str):
        self.api_url = "https://api-inference.huggingface.co/models/"
        self.headers = {"Authorization": f"Bearer {api_token}"}

    def query_model(self, model_name: str, payload: dict):
        """
        Send inference request to Hugging Face hosted model.
        Args:
            model_name (str): Hugging Face model identifier (e.g., 'facebook/prophet').
            payload (dict): Input data for inference.
        Returns:
            dict: Parsed JSON response or None if error.
        """
        try:
            response = requests.post(
                self.api_url + model_name,
                headers=self.headers,
                json=payload,
                timeout=10
            )
            if response.status_code == 200:
                logger.info(f"Model {model_name} inference successful.")
                return response.json()
            else:
                logger.error(f"Error {response.status_code} from {model_name}: {response.text}")
                return None
        except Exception as e:
            logger.error(f"Exception querying {model_name}: {e}")
            return None

    def batch_query(self, model_list: list, payload: dict):
        """
        Query multiple models in sequence (can later parallelize).
        Args:
            model_list (list): List of Hugging Face model identifiers.
            payload (dict): Input data for inference.
        Returns:
            dict: Mapping of model_name -> output.
        """
        results = {}
        for model in model_list:
            results[model] = self.query_model(model, payload)
        return results
```

```python
"""
meta_learner.py
Module for aggregating ensemble forecasts in Weather Aggregator.
Implements meta-learning strategies (ridge, logistic, boosting) and confidence band estimation.
"""

import numpy as np
from sklearn.linear_model import Ridge, LogisticRegression, LinearRegression
from sklearn.ensemble import GradientBoostingRegressor

class MetaLearner:
    def __init__(self, method="ridge"):
        """
        Initialize meta learner.
        Args:
            method (str): 'ridge', 'logistic', 'linear', or 'boosting'
        """
        if method == "ridge":
            self.model = Ridge(alpha=1.0)
        elif method == "logistic":
            self.model = LogisticRegression(max_iter=500)
        elif method == "linear":
            self.model = LinearRegression()
        elif method == "boosting":
            self.model = GradientBoostingRegressor(n_estimators=100)
        else:
            raise ValueError("Unsupported meta learner method")

    def fit(self, X, y):
        """
        Fit meta learner on ensemble outputs.
        Args:
            X (np.ndarray): shape (n_samples, n_models), model outputs
            y (np.ndarray): target values (e.g., observed weather)
        """
        self.model.fit(X, y)

    def predict(self, X):
        """
        Predict aggregated forecast.
        Args:
            X (np.ndarray): shape (n_samples, n_models), model outputs
        Returns:
            np.ndarray: aggregated forecast
        """
        return self.model.predict(X)

    def confidence_band(self, forecasts, alpha=0.1):
        """
        Compute confidence band around forecasts.
        Args:
            forecasts (np.ndarray): ensemble forecasts
            alpha (float): confidence level (default 0.1 ~ 90% band)
        Returns:
            tuple: (lower_bound, upper_bound)
        """
        mean_forecast = np.mean(forecasts, axis=1)
        std_dev = np.std(forecasts, axis=1)
        lower = mean_forecast - (1.96 * std_dev)
        upper = mean_forecast + (1.96 * std_dev)
        return lower, upper
```

- **Input**: Predictions from local models (`base_models.py`) and Hugging Face models (`huggingface_models.py`).
- **Output**: Aggregated forecast + confidence bands.
- **Flexibility**: Switch between ridge, logistic, linear, or boosting meta learners depending on interpretability vs. performance.
- **Governance**: Confidence bands provide transparency and uncertainty quantification.

```python
Includes SARIMAX, Kalman filter, Random Forest, CNN/LSTM stubs, and Autoencoder.

import numpy as np
import pandas as pd
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Optional deep learning stubs
import torch

# -------------------------------
# Classical Models
# -------------------------------

def sarimax_forecast(series, steps=5):
    """Fit SARIMAX and forecast."""
    model = SARIMAX(series, order=(1,1,1), seasonal_order=(0,0,0,0))
    fit = model.fit(disp=False)
    forecast = fit.forecast(steps=steps)
    return forecast

def kalman_filter_forecast(series, steps=5):
    """Simple Kalman filter stub (moving average as proxy)."""
    # In practice: use pykalman or custom state-space model
    smoothed = pd.Series(series).rolling(window=3, min_periods=1).mean().values
    forecast = np.tile(smoothed[-1], steps)
    return forecast

# -------------------------------
# Machine Learning Models
# -------------------------------

def random_forest_forecast(series, steps=5):
    """Random Forest regression forecast."""
    X = np.arange(len(series)).reshape(-1,1)
    y = series
    rf = RandomForestRegressor(n_estimators=50, random_state=42)
    rf.fit(X, y)
    future_X = np.arange(len(series), len(series)+steps).reshape(-1,1)
    forecast = rf.predict(future_X)
    return forecast

# -------------------------------
# Deep Learning Stubs
# -------------------------------

class SimpleLSTM(nn.Module):
    """Minimal LSTM for sequence forecasting."""
    def __init__(self, input_size=1, hidden_size=50, output_size=1):
        super(SimpleLSTM, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, batch_first=True)
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        out, _ = self.lstm(x)
        out = self.fc(out[:, -1, :])
        return out

def lstm_forecast_stub(series, steps=5):
    """Stub for LSTM forecast (returns repeated last value)."""
    return np.tile(series[-1], steps)

def autoencoder_stub(series, steps=5):
    """Stub for anomaly detection via autoencoder (returns mean)."""
    mean_val = np.mean(series)
    return np.tile(mean_val, steps)
```

```python
"""

LiteSQL (SQLite) interface for Weather Aggregator.
Handles user input storage, forecast persistence, and audit logging.
"""

import sqlite3
import os
from datetime import datetime

DB_PATH = "weather_data.db"

class DataStore:
    def __init__(self, db_path=DB_PATH):
        self.db_path = db_path
        self._init_db()

    def _init_db(self):
        """Initialize database tables if they don't exist."""
        conn = sqlite3.connect(self.db_path)
        cursor = conn.cursor()

        # User input table
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS user_input (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                po_box TEXT,
                city TEXT,
                country TEXT,
                timestamp DATETIME DEFAULT CURRENT_TIMESTAMP
            )
        ''')

        # Forecast results table
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS forecasts (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                city TEXT,
                country TEXT,
                model_name TEXT,
                forecast TEXT,
                lower_band TEXT,
                upper_band TEXT,
                timestamp DATETIME DEFAULT CURRENT_TIMESTAMP
            )
        ''')

        # Audit log table
        cursor.execute('''
            CREATE TABLE IF NOT EXISTS audit_log (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                action TEXT,
                details TEXT,
                timestamp DATETIME DEFAULT CURRENT_TIMESTAMP
            )
        ''')

        conn.commit()
        conn.close()

    def save_user_input(self, po_box, city, country):
        """Save user input into database."""
        conn = sqlite3.connect(self.db_path)
        cursor = conn.cursor()
        cursor.execute(
            'INSERT INTO user_input (po_box, city, country) VALUES (?, ?, ?)',
            (po_box, city, country)
        )
        conn.commit()
        conn.close()

    def save_forecast(self, city, country, model_name, forecast, lower_band, upper_band):
        """Save forecast results into database."""
        conn = sqlite3.connect(self.db_path)
        cursor = conn.cursor()
        cursor.execute(
            'INSERT INTO forecasts (city, country, model_name, forecast, lower_band, upper_band) VALUES (?, ?, ?, ?, ?, ?)',
            (city, country, model_name, str(forecast), str(lower_band), str(upper_band))
        )
        conn.commit()
        conn.close()

    def log_action(self, action, details=""):
        """Log governance/audit actions."""
        conn = sqlite3.connect(self.db_path)
        cursor = conn.cursor()
        cursor.execute(
            'INSERT INTO audit_log (action, details) VALUES (?, ?)',
            (action, details)
        )
        conn.commit()
        conn.close()

    def get_forecasts(self, city, country):
        """Retrieve forecasts for a given city/country."""
        conn = sqlite3.connect(self.db_path)
        cursor = conn.cursor()
        cursor.execute(
            'SELECT model_name, forecast, lower_band, upper_band, timestamp FROM forecasts WHERE city=? AND country=?',
            (city, country)
        )
        rows = cursor.fetchall()
        conn.close()
        return rows
```

```python
    Returns:
        logging.Logger: Configured logger instance.
    """
    logger = logging.getLogger(name)
    logger.setLevel(logging.INFO)

    # File handler
    fh = logging.FileHandler(LOG_FILE)
    fh.setLevel(logging.INFO)

    # Console handler
    ch = logging.StreamHandler()
    ch.setLevel(logging.INFO)

    # Formatter
    formatter = logging.Formatter(
        "%(asctime)s - %(name)s - %(levelname)s - %(message)s"
    )
    fh.setFormatter(formatter)
    ch.setFormatter(formatter)

    # Avoid duplicate handlers
    if not logger.handlers:
        logger.addHandler(fh)
        logger.addHandler(ch)

    return logger
```

```python
"""
explainability.py
Explainability utilities for Weather Aggregator.
Provides SHAP and LIME analysis for ensemble forecasts.
"""

import numpy as np
import shap
from lime.lime_tabular import LimeTabularExplainer

class Explainability:
    def __init__(self, training_data, feature_names=None):
        """
        Initialize explainability tools.
        Args:
            training_data (np.ndarray): Data used to train models.
            feature_names (list): Optional list of feature names.
        """
        self.training_data = training_data
        self.feature_names = feature_names if feature_names else [f"f{i}" for i in range(training_data.shape[1])]

    def shap_analysis(self, model, sample):
        """
        Run SHAP analysis on a given model and sample.
        Args:
            model: Trained model (supports predict or predict_proba).
            sample (np.ndarray): Single input sample.
        Returns:
            shap_values: SHAP values for the sample.
        """
        explainer = shap.Explainer(model, self.training_data)
        shap_values = explainer(sample)
        return shap_values

    def lime_analysis(self, model, sample):
        """
        Run LIME analysis on a given model and sample.
        Args:
            model: Trained model (supports predict).
            sample (np.ndarray): Single input sample.
        Returns:
            explanation: LIME explanation object.
        """
        explainer = LimeTabularExplainer(
            training_data=self.training_data,
            feature_names=self.feature_names,
            verbose=True,
            mode="regression"
        )
        explanation = explainer.explain_instance(sample, model.predict, num_features=5)
        return explanation
```

## 📓 Parallel Query Extension

```python
# Cell 6: Parallel Query with concurrent.futures
import concurrent.futures

def parallel_query(models, payload, max_workers=5):
    """
    Query multiple Hugging Face models in parallel.
    Args:
        models (list): List of model identifiers.
        payload (dict): Input data for inference.
        max_workers (int): Number of parallel workers.
    Returns:
        dict: Mapping of model_name -> output.
    """
    results = {}

    def worker(model):
        return model, query_hf_model(model, payload)

    with concurrent.futures.ThreadPoolExecutor(max_workers=max_workers) as executor:
        future_to_model = {executor.submit(worker, model): model for model in models}
        for future in concurrent.futures.as_completed(future_to_model):
            model = future_to_model[future]
            try:
                model_name, output = future.result()
                results[model_name] = output
            except Exception as e:
                results[model] = None
                logger.error(f"Parallel query failed for {model}: {e}")
    return results

# Example usage
input_sequence = "Temperature trend: 20, 21, 22, 23, 24"
parallel_results = parallel_query(MODEL_LIST, {"inputs": input_sequence}, max_workers=5)
parallel_results
```

```python
# Cell 7: Asyncio Alternative (for advanced scaling)
import asyncio
import aiohttp

async def async_query_model(session, model, payload):
    url = HF_API_URL + model
    async with session.post(url, headers=headers, json=payload) as response:
        if response.status == 200:
            return model, await response.json()
        else:
            logger.error(f"Error {response.status} from {model}")
            return model, None

async def async_batch_query(models, payload):
    async with aiohttp.ClientSession() as session:
        tasks = [async_query_model(session, model, payload) for model in models]
        results = await asyncio.gather(*tasks)
        return dict(results)

# Example usage
payload = {"inputs": "Temperature trend: 20, 21, 22, 23, 24"}
async_results = asyncio.run(async_batch_query(MODEL_LIST, payload))
```

```python
        future_to_model = {executor.submit(worker, model): model for model in models}
        for future in concurrent.futures.as_completed(future_to_model):
            model, output = future.result()
            results[model] = output
    return results

start = time.time()
par_results = parallel_query(MODEL_LIST, payload, max_workers=5)
par_time = time.time() - start
print(f"Parallel time: {par_time:.2f}s")
```

```python
# Cell 4: Async Query (aiohttp)
async def async_query_model(session, model, payload):
    url = HF_API_URL + model
    async with session.post(url, headers=headers, json=payload) as response:
        return model, await response.json() if response.status == 200 else (model, None)

async def async_batch_query(models, payload):
    async with aiohttp.ClientSession() as session:
        tasks = [async_query_model(session, model, payload) for model in models]
        results = await asyncio.gather(*tasks)
        return dict(results)

start = time.time()
async_results = asyncio.run(async_batch_query(MODEL_LIST, payload))
async_time = time.time() - start
print(f"Async time: {async_time:.2f}s")
```

- Go to [huggingface.co](https://huggingface.co).
- Click **Sign Up** (top right).
- Register using your email, GitHub, or Google account.
- Verify your email address (this step is required before tokens can be created).

---

### 2. Navigate to Access Tokens
- Once logged in, click your **profile picture** (top right).
- Select **Settings**.
- In the left sidebar, choose **Access Tokens**.

---

### 3. Generate a new token
- Click **New Token**.
- Give it a descriptive name (e.g., `weather_app`).
- Choose a **role/scope**:
  - **Read** → allows downloading models/datasets (most common for inference).
  - **Write** → allows uploading models/datasets.
  - **Admin** → full access (use cautiously).
- Click **Generate**.

---

### 4. Copy and store the token securely
- The token will look like:  
  ```
  hf_xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx
  ```
- Copy it and store it in a safe place.
- You’ll use it in your app like:
  ```python
  import os
  from huggingface_hub import InferenceApi

  HF_TOKEN = os.getenv("HF_TOKEN")  # or paste directly
  api = InferenceApi(repo_id="bert-base-uncased", token=HF_TOKEN)
  ```

---

### 5. Best practices
- **Do not hardcode** the token in public repos.
- Use environment variables (`HF_TOKEN`) or a `.env` file.
- Rotate tokens periodically for security.

````python
# Jupyter cell: recursively convert all .py files in folder + subfolders to .txt

import os

# Root directory to scan
root_folder = "."

for dirpath, dirnames, filenames in os.walk(root_folder):
    for filename in filenames:
        if filename.endswith(".py"):
            py_path = os.path.join(dirpath, filename)
            txt_path = os.path.join(dirpath, filename.replace(".py", ".txt"))
            
            # Read Python file
            with open(py_path, "r", encoding="utf-8") as f:
                code = f.read()
            
            # Write to text file
            with open(txt_path, "w", encoding="utf-8") as f:
                f.write(code)
            
            print(f"Converted: {py_path} → {txt_path}")

print("Recursive batch conversion complete.")
````

````python
# Jupyter cell: recursively convert .css, .html, .js files to .txt

import os

# Root directory to scan
root_folder = "."

# File extensions to convert
extensions = [".css", ".html", ".js"]

for dirpath, dirnames, filenames in os.walk(root_folder):
    for filename in filenames:
        if any(filename.endswith(ext) for ext in extensions):
            src_path = os.path.join(dirpath, filename)
            txt_path = os.path.join(dirpath, filename + ".txt")  # append .txt
            
            # Read source file
            with open(src_path, "r", encoding="utf-8") as f:
                content = f.read()
            
            # Write to text file
            with open(txt_path, "w", encoding="utf-8") as f:
                f.write(content)
            
            print(f"Converted: {src_path} → {txt_path}")

print("Recursive conversion complete.")
````
