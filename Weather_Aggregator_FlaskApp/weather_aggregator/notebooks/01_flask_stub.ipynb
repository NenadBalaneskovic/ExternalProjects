Stub Flask App: Minimal app with inline template.

Input Form: Captures PO Box, City, Country.

Prototype Output: Displays submitted values back to user.

Narrative: Demonstrates the frontend flow before moving to modular app/ structure.

```python
# Cell 1: Imports & Setup
from flask import Flask, render_template_string, request

# Minimal Flask app stub
app = Flask(__name__)
app.config["SECRET_KEY"] = "stubsecret"
```

```python
# Cell 2: Inline Template (Prototype)
index_template = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>Weather Aggregator Stub</title>
</head>
<body>
    <h1>Weather Aggregator (Stub)</h1>
    <form method="POST">
        <label>PO Box:</label><br>
        <input type="text" name="po_box"><br><br>
        <label>City:</label><br>
        <input type="text" name="city"><br><br>
        <label>Country:</label><br>
        <input type="text" name="country"><br><br>
        <button type="submit">Submit</button>
    </form>
    {% if result %}
    <h2>Submitted Data</h2>
    <p>PO Box: {{ result.po_box }}</p>
    <p>City: {{ result.city }}</p>
    <p>Country: {{ result.country }}</p>
    {% endif %}
</body>
</html>
"""
```

```python
# Cell 3: Routes (Stub)
@app.route("/", methods=["GET", "POST"])
def index():
    result = None
    if request.method == "POST":
        result = {
            "po_box": request.form.get("po_box"),
            "city": request.form.get("city"),
            "country": request.form.get("country")
        }
    return render_template_string(index_template, result=result)
```
