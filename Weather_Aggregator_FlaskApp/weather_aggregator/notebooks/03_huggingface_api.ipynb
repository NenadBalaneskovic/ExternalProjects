✅ What This Notebook Achieves  

API Integration: Demonstrates querying Hugging Face hosted models.

Batch Query: Shows how to query multiple models in sequence.

Explainability Stub: Provides narratable outputs for consulting demos.

Narrative: Bridges hosted inference with ensemble explainability.

```python
# Cell 1: Imports & Setup
import requests
import json

# Hugging Face API endpoint and token (replace with your own)
HF_API_URL = "https://api-inference.huggingface.co/models/"
HF_API_TOKEN = "your_huggingface_token_here"

headers = {"Authorization": f"Bearer {HF_API_TOKEN}"}
```

```python
# Cell 2: Query Function
def query_hf_model(model_name, payload):
    """Query Hugging Face hosted model."""
    response = requests.post(HF_API_URL + model_name, headers=headers, json=payload)
    if response.status_code == 200:
        return response.json()
    else:
        print(f"Error {response.status_code}: {response.text}")
        return None
```

```python
# Cell 3: Example Forecast Query (Stub)
# Using a text-generation model as placeholder
payload = {"inputs": "Weather forecast for Frankfurt tomorrow"}
result = query_hf_model("gpt2", payload)

print("Hosted Model Result:", json.dumps(result, indent=2))
```

```python
# Cell 4: Batch Query
def batch_query(models, payload):
    results = {}
    for model in models:
        results[model] = query_hf_model(model, payload)
    return results

models = ["gpt2", "distilgpt2"]
batch_results = batch_query(models, {"inputs": "Temperature trend in Frankfurt"})
print(batch_results)
```

```python
# Cell 5: Narratable Explainability Stub
# In practice, you would wrap hosted outputs with SHAP/LIME explainability
def narrate_results(results):
    for model, output in results.items():
        print(f"Model {model} produced output: {output}")

narrate_results(batch_results)
```
