✅ What This Notebook Achieves  

Synthetic Data: Creates a simple temperature series.

Base Models: SARIMAX, Kalman, Random Forest forecasts.

Meta Learner: Ridge regression aggregation.

Visualization: Compares base models and ensemble forecast.

Narrative: Demonstrates ensemble prototyping for consulting demos.

```python
# Cell 1: Imports & Setup
import numpy as np
import statsmodels.api as sm
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Ridge
import matplotlib.pyplot as plt
```

```python
# Cell 2: Synthetic Data
np.random.seed(42)
time = np.arange(50)
series = 20 + 0.1*time + np.random.normal(scale=1.0, size=50)

plt.plot(time, series, label="Synthetic Temperature Series")
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

```python
# Cell 3: SARIMAX Forecast
sarimax = sm.tsa.SARIMAX(series, order=(1,1,1))
sarimax_fit = sarimax.fit(disp=False)
sarimax_forecast = sarimax_fit.forecast(steps=5)

print("SARIMAX Forecast:", sarimax_forecast)
```

```python
# Cell 4: Kalman Filter Forecast (Stub)
def kalman_filter_forecast(series, steps=5):
    return [np.mean(series[-3:])] * steps

kalman_forecast = kalman_filter_forecast(series, steps=5)
print("Kalman Forecast:", kalman_forecast)
```

```python
# Cell 5: Random Forest Forecast
X = np.arange(len(series)).reshape(-1,1)
y = series
rf = RandomForestRegressor(n_estimators=50, random_state=42)
rf.fit(X, y)
future = np.arange(len(series), len(series)+5).reshape(-1,1)
rf_forecast = rf.predict(future)

print("Random Forest Forecast:", rf_forecast)
```

```python
# Cell 6: Meta Learner Aggregation
X_meta = np.vstack([sarimax_forecast, kalman_forecast, rf_forecast]).T
y_meta = np.mean(X_meta, axis=1)

meta = Ridge(alpha=1.0)
meta.fit(X_meta, y_meta)
ensemble_forecast = meta.predict(X_meta)

print("Ensemble Forecast:", ensemble_forecast)
```

```python
# Cell 7: Visualization
plt.figure(figsize=(10,6))
plt.plot(range(len(series)), series, label="Observed Series")
plt.plot(range(len(series), len(series)+5), sarimax_forecast, label="SARIMAX")
plt.plot(range(len(series), len(series)+5), kalman_forecast, label="Kalman")
plt.plot(range(len(series), len(series)+5), rf_forecast, label="Random Forest")
plt.plot(range(len(series), len(series)+5), ensemble_forecast, label="Ensemble", linewidth=2, color="black")
plt.legend()
plt.title("Model Development: Forecast Comparison")
plt.show()
```
