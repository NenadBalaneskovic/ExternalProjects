✅ What This Notebook Achieves  

Forecast Visualization: Plots forecasts with confidence bands for each model.

Ensemble Aggregation: Shows meta learner output alongside base models.

Reporting Stub: Prints a simple governance‑style report.

Narrative: Provides a dashboard view that executives can interpret easily.

```python
# Cell 1: Imports & Setup
import numpy as np
import matplotlib.pyplot as plt

# Synthetic ensemble forecasts
models = ["SARIMAX", "RandomForest", "Kalman"]
forecast_values = {
    "SARIMAX": [22.5, 23.0, 21.8, 22.2, 23.1],
    "RandomForest": [22.0, 22.7, 23.1, 22.8, 23.3],
    "Kalman": [22.3, 22.4, 22.5, 22.6, 22.7]
}
confidence_bands = {
    "SARIMAX": ([21.0, 21.5, 21.2, 21.7, 22.0], [24.0, 24.5, 24.2, 24.7, 25.0]),
    "RandomForest": ([21.5, 21.8, 22.0, 22.1, 22.5], [23.5, 23.8, 24.0, 24.1, 24.5]),
    "Kalman": ([22.0, 22.1, 22.2, 22.3, 22.4], [22.6, 22.7, 22.8, 22.9, 23.0])
}
```

```python
# Cell 2: Forecast Visualization
plt.figure(figsize=(10,6))

for model in models:
    forecast = forecast_values[model]
    lower, upper = confidence_bands[model]
    plt.plot(forecast, label=f"{model} Forecast")
    plt.fill_between(range(len(forecast)), lower, upper, alpha=0.2)

plt.title("Weather Forecasts with Confidence Bands")
plt.xlabel("Time Steps")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

```python
# Cell 3: Ensemble Aggregation (Stub)
from sklearn.linear_model import Ridge

X = np.array([forecast_values[m] for m in models]).T
y = np.mean(X, axis=1)  # proxy target
meta = Ridge(alpha=1.0)
meta.fit(X, y)
ensemble_forecast = meta.predict(X)

plt.figure(figsize=(10,6))
plt.plot(ensemble_forecast, label="Ensemble Forecast", color="black", linewidth=2)
plt.title("Ensemble Forecast (Meta Learner)")
plt.xlabel("Time Steps")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
```

```python
# Cell 4: Reporting Stub
def generate_report(city, country, forecasts, confidence_bands):
    print(f"Weather Report for {city}, {country}")
    for model, forecast in forecasts.items():
        lb, ub = confidence_bands[model]
        print(f"{model}: {forecast} | Band=({lb[0]}, {ub[-1]})")

generate_report("Frankfurt", "Germany", forecast_values, confidence_bands)
```
